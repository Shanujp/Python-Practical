```python
inputString=input("Enter your name: ").split()
print ("Input sentence is: ", inputString)
remDuplicate=[]
for i in inputString:
    if i not in remDuplicate:
        remDuplicate.append(i)

print("Unique words are: ", remDuplicate)
for i in remDuplicate:
    print(i," : " ,inputString.count(i), end=" ")
```
