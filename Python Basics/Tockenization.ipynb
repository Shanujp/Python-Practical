```python
## test preprocessing stemming using NLTK
## Comments of product is a positive review or negative review
# porter stemmer
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
words = ["playing", "Played", "plays", "runner", "running", "runs","happiness"]
for word in words:
    print(stemming.stem(word))

stemming.stem("happiness")
```

```python
stemming.stem("happiness")
## Disadvantage of porter stemmer is that it will not give the correct root word, but it will give the root word with some extra characters. For example, happiness will be stemmed to happi, which is not a correct root word.
```

```python
# regex stemmer
from nltk.stem import RegexpStemmer
stemmer = RegexpStemmer('(ing|ed|s|ness)$') ##ending with ing, ed, s, ness will removed
## $ing means remove the starting with ing, ed, s, ness will not be removed
for word in words:
    print(stemming.stem(word))
```

```python
##Snowball stemmer
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer("english")
print(snowball_stemmer.stem("happiness"))
```

```python
## Lemmatization
from nltk.stem import WordNetLemmatizer
##WordNetLemmatizer only for handle plura, tense 
lemmatizer = WordNetLemmatizer() 
for word in words:
    print(lemmatizer.lemmatize(word,pos="v"))
```

```python
#stoping words
##top words are very common words that do not add important meaning to a sentence.
##Reduce text size
# Faster processing
# Better ML accuracy
# Remove unnecessary noise
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
sentence = "I am learning Natural Language Processing in Python"

stop_words = set(stopwords.words('english'))

words = word_tokenize(sentence)

filtered_words = [w for w in words if w.lower() not in stop_words]

print(filtered_words)
```

```python
from nltk.tokenize import sent_tokenize
from nltk.corpus import stopwords

from nltk.stem import WordNetLemmatizer

paragrahph = "I am learning Natural Language Processing in Python. It is a very interesting field. I hope to learn more about it in the future."
stop_words = set(stopwords.words('english'))
# step 1 conver paragraph into sentences
sentences = sent_tokenize(paragrahph)
stemmedStentences = []
lemnatizedSentences = []
lemmatizer = WordNetLemmatizer() 
stemming = PorterStemmer()
# step 2 apply stopword and filter and then apply stemming
for i in range(len(sentences)):
    # word tokenization : convert sentence into words
    words = word_tokenize(sentences[i])
    filtered_words = [w for w in words if w.lower() not in stop_words]  ## remove stop words
    lemmatized_words = [lemmatizer.lemmatize(w) for w in filtered_words]
    lemnatizedSentences.append(" ".join( lemmatized_words))
    stemmedStentences.append(" ".join( [stemming.stem(w) for w in filtered_words]))
print("Lementized senetece : " + "\n".join(lemnatizedSentences))
print("Stemmed sentences : " + "\n".join(stemmedStentences))
```
